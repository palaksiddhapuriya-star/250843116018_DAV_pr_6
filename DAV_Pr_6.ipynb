Practical_6: Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={"Month":["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"],
      "Product":["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet"],
      "Region": ["North","South","East","West","North","South","East","West","North","South","East","West"],
      "sales":[120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
      "Units_sold":[12,30,25,14,35,30,18,38,32,21,45,40]}

df = pd.DataFrame(data)
df
```

```python
df.head()
```

```python
# Task 2: Line Chart - Sales from Jan to Dec
plt.figure(figsize=(10,5))
plt.plot(df["Month"], df["sales"], marker="o")
plt.title("Monthly Sales from January to December")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

```python
# Task 3 — Bar Chart: Total sales by Product
product_sales = df.groupby("Product")["sales"].sum()

product_sales.plot(kind="bar", figsize=(7,5))
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
# Task 4 — Histogram: Distribution of individual sales values

plt.figure(figsize=(8,5))
plt.hist(df["sales"], bins=6, edgecolor="black")
plt.title("Distribution of Sales Values")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
# Task 5: Scatter Plot - Units Sold vs Sales
plt.figure(figsize=(8,5))
plt.scatter(df["Units_sold"], df["sales"])
plt.title("Relationship Between Units Sold and Total Sales")
plt.xlabel("Units Sold")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

```python
# Task 6: Pie Chart - Sales Percentage by Region
region_sales = df.groupby("Region")["sales"].sum()

plt.figure(figsize=(8, 5))
plt.pie(region_sales, labels=region_sales.index, autopct="%1.1f%%")
plt.title("Sales Percentage by Region")
plt.show()
```
